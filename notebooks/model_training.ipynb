```python
import pandas as pd
```

```python
df1=pd.read_csv("with_pca.csv")
df2=pd.read_csv("without_pca.csv")
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.svm import SVC
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.ensemble import GradientBoostingClassifier
from sklearn.ensemble import AdaBoostClassifier
from xgboost import XGBClassifier
from sklearn.model_selection import RandomizedSearchCV
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, roc_auc_score
```

```python
x1=df1[[i for i in df1.columns if i !='class']]
y1=df1['class']
x2=df2[[i for i in df2.columns if i !='class']]
y2=df2['class']
```

```python
x1_test, x1_train, y1_test, y1_train = train_test_split(x1, y1, test_size=0.2, random_state=42)
x2_test, x2_train, y2_test, y2_train = train_test_split(x1, y1, test_size=0.2, random_state=42)
```

```python
knn_param={'n_neighbors':[3,5,7,9,11,13,15],'weights':['uniform','distance'],'algorithm':['auto','ball_tree','kd_tree','brute']}
lr_param={'penalty':['l1','l2','elasticnet'],'C':[0.01,0.1,1,10,100],'solver':['newton-cg','lbfgs','liblinear','sag','saga']}
svm_param={'C':[0.01,0.1,1,10,100],'kernel':['linear','poly','rbf','sigmoid'],'gamma':['scale','auto']}
dt_param={'criterion':['gini','entropy'],'max_depth':[None,5,10,15,20],'min_samples_split':[2,5,10],'min_samples_leaf':[1,2,4]}
gb_param={'n_estimators':[50,100,200],'learning_rate':[0.1,0.01,0.001],'max_depth':[3,5,7]}
rf_param={'n_estimators':[50,100,200],'max_depth':[None,5,10,15],'min_samples_split':[2,5,10],'min_samples_leaf':[1,2,4]}
xgb_param={'n_estimators':[50,100,200],'learning_rate':[0.1,0.01,0.001],'max_depth':[3,5,7]}
ad_param={'n_estimators':[50,100,200],'learning_rate':[0.1,0.01,0.001]}
grid_params={'knn':knn_param,'lr':lr_param,'svm':svm_param,'dt':dt_param,'gb':gb_param,'rf':rf_param,'xgb':xgb_param,'ad':ad_param}
models={'knn':KNeighborsClassifier(),'lr':LogisticRegression(),'svm':SVC(),'dt':DecisionTreeClassifier(),'gb':GradientBoostingClassifier(),'rf':RandomForestClassifier(),'xgb':XGBClassifier(),'ad':AdaBoostClassifier()}
```

```python
import warnings
warnings.filterwarnings('ignore')
```

```python
i=0
for name,params in grid_params.items():
    model=models[name]
    i+=1
    gr=RandomizedSearchCV(model,params,cv=5,n_iter=10,scoring='accuracy')
    gr.fit(x1_train,y1_train)
    gr_pred=gr.predict(x1_test)
    print(f"Model: {model}")
    print(f"Best Parameters: {gr.best_params_}")
    print(f"Accuracy: {accuracy_score(y1_test,gr_pred)}")    
    print(f"Precision: {precision_score(y1_test,gr_pred)}")
    print(f"Recall: {recall_score(y1_test,gr_pred)}")
    print(f"F1 Score: {f1_score(y1_test,gr_pred)}")
    print(f"ROC AUC Score: {roc_auc_score(y1_test,gr_pred)}")
```

BEST MODEL WITH PCA : XGBOOST

```python
i=0
for name,params in grid_params.items():
    model=models[name]
    i+=1
    gr=RandomizedSearchCV(model,params,cv=5,n_iter=10,scoring='accuracy')
    gr.fit(x2_train,y1_train)
    gr_pred=gr.predict(x2_test)
    print(f"Model: {model}")
    print(f"Best Parameters: {gr.best_params_}")
    print(f"Accuracy: {accuracy_score(y2_test,gr_pred)}")    
    print(f"Precision: {precision_score(y2_test,gr_pred)}")
    print(f"Recall: {recall_score(y2_test,gr_pred)}")
    print(f"F1 Score: {f1_score(y2_test,gr_pred)}")
    print(f"ROC AUC Score: {roc_auc_score(y2_test,gr_pred)}")
```

BEST MODEL WITHOUT PCA : XGBOOST

THE ACCURACY,PRECISION AND RECALL OF BOTH DATASET BEFORE AND AFTER PCA ARE APPROXIMATELY SAME FOR BEST MODELS. BUT FOR OTHER MODELS THE PERFORMANCE METRICS HAVE IMPROVED LIKE 1-3 PERCENT.

WITH PCA

```python
model=XGBClassifier(n_estimators= 200, max_depth= 3, learning_rate= 0.1)
model.fit(x1_train,y1_train)
gr_pred=model.predict(x1_test)
print(f"Accuracy: {accuracy_score(y1_test,gr_pred)}")    
print(f"Precision: {precision_score(y1_test,gr_pred)}")
print(f"Recall: {recall_score(y1_test,gr_pred)}")
print(f"F1 Score: {f1_score(y1_test,gr_pred)}")
print(f"ROC AUC Score: {roc_auc_score(y1_test,gr_pred)}")
```

```python
model=XGBClassifier(n_estimators= 200, max_depth= 3, learning_rate= 0.1)
model.fit(x2_train,y2_train)
gr_pred=model.predict(x2_test)
print(f"Accuracy: {accuracy_score(y2_test,gr_pred)}")    
print(f"Precision: {precision_score(y2_test,gr_pred)}")
print(f"Recall: {recall_score(y2_test,gr_pred)}")
print(f"F1 Score: {f1_score(y2_test,gr_pred)}")
print(f"ROC AUC Score: {roc_auc_score(y2_test,gr_pred)}")
```
